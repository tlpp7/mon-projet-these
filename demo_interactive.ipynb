## 🔬 1. Algorithme de Sinkhorn-JKO (Transport de Masse)
Calcul du coût de déplacement d'une distribution de mesure sur les géodésiques de la variété.

```python
n = 80
x = np.linspace(-2, 2, n)
X, Y = np.meshgrid(x, x)
C_matrix = (X - Y)**2
epsilon = 0.01
K_matrix = np.exp(-C_matrix / epsilon)

plt.figure(figsize=(5, 4))
plt.imshow(K_matrix, cmap='viridis')
plt.title("Noyau de Transport de Wasserstein Regularisé")
plt.colorbar()
plt.show()
```
